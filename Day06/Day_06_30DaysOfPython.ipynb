## ১. টাপল কী এবং কীভাবে তৈরি করতে হয়? (Tuples & Creating a Tuple)

**Tuple (টাপল)** হলো বিভিন্ন ডেটা টাইপের একটি **Ordered** (ক্রমবদ্ধ) এবং **Unchangeable / Immutable** (অপরিবর্তনযোগ্য) কালেকশন। টাপল প্রথম বন্ধনী বা রাউন্ড ব্র্যাকেট **`()`** দিয়ে লেখা হয়।
- একবার টাপল তৈরি হয়ে গেলে এর ভেতরের মান পরিবর্তন করা যায় না।
- যেহেতু টাপল পরিবর্তনযোগ্য (Mutable) নয়, তাই এতে `add`, `insert` বা `remove` মেথড ব্যবহার করা যায় না।

### টাপল সম্পর্কিত প্রধান মেথড ও অপারেটর:
- `tuple()` : একটি ফাঁকা (empty) টাপল তৈরি করতে।
- `count()` : টাপলের ভেতরে নির্দিষ্ট কোনো আইটেম কতবার আছে তা গণনা করতে।
- `index()` : টাপলের ভেতরে নির্দিষ্ট কোনো আইটেমের ইনডেক্স খুঁজে বের করতে।
- `+` অপারেটর : দুই বা ততোধিক টাপল জোড়া লাগিয়ে নতুন একটি টাপল তৈরি করতে।

```python
# Empty tuple: Creating an empty tuple
empty_tuple = ()
# or using the tuple constructor
empty_tuple = tuple()
print('Empty tuple:', empty_tuple)

# Tuple with initial values
tpl = ('item1', 'item2', 'item3')
fruits = ('banana', 'orange', 'mango', 'lemon')
print('Fruits:', fruits)
```

## ২. টাপলের দৈর্ঘ্য (Tuple Length)

টাপলের ভেতরে মোট কয়টি আইটেম আছে তা জানার জন্য বিল্ট-ইন **`len()`** ফাংশন ব্যবহার করা হয়।

```python
tpl = ('item1', 'item2', 'item3')
print('Length of tpl:', len(tpl))

fruits = ('banana', 'orange', 'mango', 'lemon')
print('Length of fruits:', len(fruits))
```

## ৩. টাপলের আইটেম অ্যাক্সেস করা (Accessing Tuple Items)

লিস্টের মতোই টাপলের আইটেম অ্যাক্সেস করার জন্য **Positive Indexing** এবং **Negative Indexing** ব্যবহার করা হয়:
- **Positive Indexing:** বাম দিক থেকে **`0`** ইনডেক্স দিয়ে শুরু হয় (`fruits[0]` হলো প্রথম আইটেম এবং `fruits[len(fruits) - 1]` হলো শেষ আইটেম)।
- **Negative Indexing:** ডান দিক (শেষ) থেকে **`-1`** ইনডেক্স দিয়ে শুরু হয় (`-1` হলো শেষ আইটেম, `-2` হলো শেষের আগের আইটেম এবং টাপলের দৈর্ঘ্যের ঋণাত্মক মান হলো প্রথম আইটেম)।

```python
# Positive Indexing
fruits = ('banana', 'orange', 'mango', 'lemon')
first_fruit = fruits[0]
second_fruit = fruits[1]
last_index = len(fruits) - 1
last_fruit = fruits[last_index]

print('Positive Indexing -> First:', first_fruit, '| Second:', second_fruit, '| Last:', last_fruit)

# Negative Indexing
fruits = ('banana', 'orange', 'mango', 'lemon')
first_fruit = fruits[-4]
second_fruit = fruits[-3]
last_fruit = fruits[-1]

print('Negative Indexing -> First:', first_fruit, '| Second:', second_fruit, '| Last:', last_fruit)
```

## ৪. টাপল স্লাইসিং (Slicing Tuples)

টাপলের ভেতর থেকে নির্দিষ্ট রেঞ্জের ইনডেক্স (শুরু এবং শেষ) উল্লেখ করে একটি নতুন সাব-টাপল (sub-tuple) কেটে নেওয়া যায়:
- **Range of Positive Indexes:** যেমন `fruits[1:3]` দিলে ইনডেক্স `1` থেকে শুরু হয়ে ইনডেক্স `3`-এর ঠিক আগ পর্যন্ত (অর্থাৎ `1` ও `2` ইনডেক্সের আইটেম) রিটার্ন করবে।
- **Range of Negative Indexes:** যেমন `fruits[-3:-1]` দিলে ডান দিক থেকে `-3` ইনডেক্স শুরু হয়ে `-1`-এর ঠিক আগ পর্যন্ত আইটেমগুলো রিটার্ন করবে।

```python
# Range of Positive Indexes
fruits = ('banana', 'orange', 'mango', 'lemon')
all_fruits = fruits[0:4]          # all items
all_fruits = fruits[0:]           # all items
orange_mango = fruits[1:3]        # doesn't include item at index 3
orange_to_the_rest = fruits[1:]

print('All fruits (pos):', all_fruits)
print('Orange & mango (pos):', orange_mango)
print('Orange to the rest (pos):', orange_to_the_rest)

# Range of Negative Indexes
fruits = ('banana', 'orange', 'mango', 'lemon')
all_fruits = fruits[-4:]          # all items
orange_mango = fruits[-3:-1]      # doesn't include item at index 3 (-1)
orange_to_the_rest = fruits[-3:]

print('All fruits (neg):', all_fruits)
print('Orange & mango (neg):', orange_mango)
print('Orange to the rest (neg):', orange_to_the_rest)
```

## ৫. টাপলকে লিস্টে রূপান্তর করা (Changing Tuples to Lists)

যেহেতু টাপল অপরিবর্তনযোগ্য (Immutable), তাই কোনো টাপলের মান পরিবর্তন বা মডিফাই করতে হলে প্রথমে **`list()`** ফাংশন দিয়ে সেটিকে লিস্টে রূপান্তর করতে হয়। লিস্টে পরিবর্তন করার পর আবার **`tuple()`** ফাংশন দিয়ে সেটিকে টাপলে ফিরিয়ে আনা যায়।

```python
fruits = ('banana', 'orange', 'mango', 'lemon')
fruits = list(fruits)
fruits[0] = 'apple'
print(fruits)     # ['apple', 'orange', 'mango', 'lemon']

fruits = tuple(fruits)
print(fruits)     # ('apple', 'orange', 'mango', 'lemon')
```

## ৬. টাপলে আইটেম যাচাই, জোড়া লাগানো ও ডিলিট করা (`in`, `+`, `del`)

- **Checking an Item (`in`):** টাপলের ভেতরে নির্দিষ্ট কোনো আইটেম আছে কি না তা **`in`** অপারেটর দিয়ে যাচাই করা যায়, যা বুলিয়ান (`True` বা `False`) রিটার্ন করে। সরাসরি মান বদলাতে গেলে `TypeError` দেখাবে।
- **Joining Tuples (`+`):** **`+`** অপারেটর ব্যবহার করে দুই বা ততোধিক টাপল জোড়া লাগিয়ে একটি নতুন টাপল তৈরি করা যায়।
- **Deleting Tuples (`del`):** টাপল থেকে কোনো একটি একক (single) আইটেম মুছে ফেলা সম্ভব নয়, তবে **`del`** কিওয়ার্ড ব্যবহার করে সম্পূর্ণ টাপলটিই ডিলিট করা সম্ভব।

```python
# Checking an Item in a Tuple
fruits = ('banana', 'orange', 'mango', 'lemon')
print('orange' in fruits)  # True
print('apple' in fruits)   # False
# fruits[0] = 'apple'      # TypeError: 'tuple' object does not support item assignment

# Joining Tuples
fruits = ('banana', 'orange', 'mango', 'lemon')
vegetables = ('Tomato', 'Potato', 'Cabbage', 'Onion', 'Carrot')
fruits_and_vegetables = fruits + vegetables
print('Joined tuple:', fruits_and_vegetables)

# Deleting Tuples
fruits = ('banana', 'orange', 'mango', 'lemon')
del fruits
print('Tuple deleted successfully using del!')
```
