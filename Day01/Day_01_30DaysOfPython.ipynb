## ২. গাণিতিক অপারেটরসমূহ (Arithmetic Operators)

Python-এ বিভিন্ন গাণিতিক হিসাব করার জন্য নিচের অপারেটরগুলো ব্যবহৃত হয়:

| অপারেটর | নাম | কাজ | উদাহরণ | ফলাফল |
|---|---|---|---|---|
| `+` | Addition | দুটি সংখ্যা যোগ করে | `2 + 3` | `5` |
| `-` | Subtraction | প্রথম সংখ্যা থেকে দ্বিতীয়টি বিয়োগ করে | `3 - 1` | `2` |
| `*` | Multiplication | দুটি সংখ্যা গুণ করে | `2 * 3` | `6` |
| `/` | Division | ভাগ করে এবং সবসময় দশমিক (`float`) মান দেয় | `3 / 2` | `1.5` |
| `**` | Exponentiation | ঘাত বা পাওয়ার ($a^b$) নির্ণয় করে | `3 ** 2` ($3^2$) | `9` |
| `%` | Modulus | ভাগশেষ (Remainder) নির্ণয় করে | `3 % 2` | `1` |
| `//` | Floor Division | ভাগফলের পূর্ণসংখ্যার অংশটি (দশমিক বাদ দিয়ে) দেয় | `3 // 2` | `1` |

```python
print(2 + 3)   # addition(+)
print(3 - 1)   # subtraction(-)
print(2 * 3)   # multiplication(*)
print(3 + 2)   # addition(+)
print(3 - 2)   # subtraction(-)
print(3 * 2)   # multiplication(*)
print(3 / 2)   # division(/)
print(3 ** 2)  # exponential(**)
print(3 % 2)   # modulus(%)
print(3 // 2)  # Floor division operator(//)
```

## ৩. ডেটা টাইপ পরীক্ষা করা (`type()` ফাংশন)

Python-এ কোনো ভ্যালু বা ভেরিয়েবল কোন ধরনের ডেটা তা জানার জন্য বিল্ট-ইন **`type()`** ফাংশন ব্যবহার করা হয়।

### প্রধান ডেটা টাইপসমূহ:
1. **`int` (Integer):** ধনাত্মক বা ঋণাত্মক পূর্ণসংখ্যা। যেমন: `10`, `-5`, `0`।
2. **`float` (Floating-point):** দশমিকযুক্ত সংখ্যা। যেমন: `3.14`, `9.8`।
3. **`complex`:** জটিল সংখ্যা, যেখানে বাস্তব ও কাল্পনিক (`j`) অংশ থাকে। যেমন: `1 + 3j`।
4. **`str` (String):** কোটেশনের ভেতরে থাকা যেকোনো টেক্সট বা অক্ষরের সমষ্টি। যেমন: `'Asabeneh'`।
5. **`list`:** থার্ড ব্র্যাকেট `[]`-এর ভেতরে কমা দিয়ে সাজানো পরিবর্তনযোগ্য (mutable) ও ক্রমবদ্ধ (ordered) ডেটার তালিকা। যেমন: `[1, 2, 3]`।
6. **`dict` (Dictionary):** কার্লি ব্রেস `{}`-এর ভেতরে `key: value` জোড়ায় সাজানো ডেটা। যেমন: `{'name': 'Asabeneh'}`।
7. **`set`:** কার্লি ব্রেস `{}`-এর ভেতরে কমা দিয়ে রাখা অদ্বিতীয় (unique) ও অক্রমবদ্ধ (unordered) উপাদানের সংগ্রহ। যেমন: `{9.8, 3.14, 2.7}`।
8. **`tuple`:** প্রথম বন্ধনী `()`-এর ভেতরে কমা দিয়ে সাজানো অপরিবর্তনযোগ্য (immutable) ডেটার তালিকা। যেমন: `(9.8, 3.14, 2.7)`।

> **গুরুত্বপূর্ণ সংশোধন (Note on Set vs Tuple):**
> আপনার দেওয়া কোডের শেষ লাইনে `print(type({9.8, 3.14, 2.7}))`-এর পাশে কমেন্টে `# Tuple` লেখা ছিল। কিন্তু Python-এ কার্লি ব্রেস `{...}` দিয়ে লিখলে সেটি **`set`** (`<class 'set'>`) হয়! **Tuple** তৈরি করতে হলে প্রথম বন্ধনী `(...)` ব্যবহার করতে হয়, যেমন: `(9.8, 3.14, 2.7)`। নিচের কোড সেলে দুটিই দেখানো হয়েছে।

```python
# Checking data types

print(type(10))                  # Int
print(type(3.14))                # Float
print(type(1 + 3j))              # Complex
print(type('Asabeneh'))          # String
print(type([1, 2, 3]))           # List
print(type({'name': 'Asabeneh'}))  # Dictionary
print(type({9.8, 3.14, 2.7}))    # Set (কার্লি ব্রেস {} থাকায় এটি Set)
print(type((9.8, 3.14, 2.7)))    # Tuple (প্রথম বন্ধনী () থাকলে সেটি Tuple)
```

## ৪. দ্রুত রিভিশন (Quick Summary)

| কোড উদাহরণ | আউটপুট টাইপ | ব্র্যাকেট / সিনট্যাক্স |
|---|---|---|
| `10` | `<class 'int'>` | পূর্ণসংখ্যা |
| `3.14` | `<class 'float'>` | দশমিক সংখ্যা |
| `1 + 3j` | `<class 'complex'>` | বাস্তব + কাল্পনিক (`j`) |
| `'Asabeneh'` | `<class 'str'>` | `'...'` বা `"..."` |
| `[1, 2, 3]` | `<class 'list'>` | Square Brackets `[]` |
| `{'name': 'Asabeneh'}` | `<class 'dict'>` | Curly Braces `{key: value}` |
| `{9.8, 3.14, 2.7}` | `<class 'set'>` | Curly Braces `{}` (শুধু ভ্যালু) |
| `(9.8, 3.14, 2.7)` | `<class 'tuple'>` | Parentheses `()` |
