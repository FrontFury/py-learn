## ১. ভেরিয়েবল কী এবং কীভাবে ডিক্লেয়ার করতে হয়? (Variables in Python)

**Variable (ভেরিয়েবল)** হলো কম্পিউটারের মেমোরিতে ডেটা বা মান সংরক্ষণ করার একটি নামযুক্ত স্থান (Container)। Python-এ ভেরিয়েবল তৈরি করার জন্য আলাদা কোনো টাইপ (যেমন: `int`, `string`) আগে থেকে বলে দিতে হয় না; সমান চিহ্ন (`=`) দিয়ে মান অ্যাসাইন করলেই ভেরিয়েবল তৈরি হয়ে যায়।

### ভেরিয়েবল নামকরণের নিয়ম (Variable Naming Rules):
- ভেরিয়েবলের নাম অবশ্যই অক্ষর (`a-z`, `A-Z`) অথবা আন্ডারস্কোর (`_`) দিয়ে শুরু হতে হবে।
- নামের শুরুতে সংখ্যা (`0-9`) ব্যবহার করা যাবে না (তবে নামের মাঝে বা শেষে সংখ্যা থাকতে পারে, যেমন: `user1`)।
- ভেরিয়েবলের নামে শুধু অক্ষর, সংখ্যা এবং আন্ডারস্কোর (`_`) থাকতে পারবে (কোনো স্পেস বা `-`, `@`, `$` ব্যবহার করা যাবে না)।
- ভেরিয়েবলের নাম **Case-sensitive** (অর্থাৎ `age`, `Age` এবং `AGE` তিনটি ভিন্ন ভেরিয়েবল)।
- Python-এ একাধিক শব্দের ভেরিয়েবল লিখতে সাধারণত **snake_case** পদ্ধতি (যেমন: `first_name`, `is_married`, `person_info`) অনুসরণ করা হয়।

```python
# Variables in Python

first_name = 'Asabeneh'
last_name = 'Yetayeh'
country = 'Finland'
city = 'Helsinki'
age = 250
is_married = True
skills = ['HTML', 'CSS', 'JS', 'React', 'Python']
person_info = {
    'firstname': 'Asabeneh',
    'lastname': 'Yetayeh',
    'country': 'Finland',
    'city': 'Helsinki'
}
```

## ২. ভেরিয়েবলের মান প্রিন্ট করা এবং `len()` ফাংশনের ব্যবহার

- **`print()` ফাংশনে একাধিক আর্গুমেন্ট:** `print()` ফাংশনের ভেতরে কমা (`,`) ব্যবহার করে টেক্সট (String) এবং ভেরিয়েবল একসাথে প্রিন্ট করা যায়। কমা ব্যবহার করলে Python স্বয়ংক্রিয়ভাবে দুটির মাঝে একটি স্পেস যোগ করে দেয়।
- **`len()` ফাংশন:** কোনো স্ট্রিং (String)-এ মোট কয়টি অক্ষর (Character) আছে অথবা কোনো লিস্ট/ডিকশনারিতে কয়টি উপাদান আছে তা গণনা করার জন্য বিল্ট-ইন `len()` ফাংশন ব্যবহৃত হয়। যেমন: `'Asabeneh'` শব্দটিতে ৮টি অক্ষর থাকায় `len(first_name)` এর মান `8`।

```python
# Printing the values stored in the variables

print('First name:', first_name)
print('First name length:', len(first_name))
print('Last name: ', last_name)
print('Last name length: ', len(last_name))
print('Country: ', country)
print('City: ', city)
print('Age: ', age)
print('Married: ', is_married)
print('Skills: ', skills)
print('Person information: ', person_info)
```

## ৩. এক লাইনে একাধিক ভেরিয়েবল ডিক্লেয়ার করা (Multiple Variable Assignment)

Python-এ কমা (`,`) ব্যবহার করে একই লাইনে একাধিক ভেরিয়েবল ডিক্লেয়ার করা এবং তাদের মান নির্ধারণ (Assign) করা যায়।

- বাম পাশের ভেরিয়েবলগুলোর ক্রম (Order) অনুযায়ী ডান পাশের মানগুলো অ্যাসাইন হয়।
- বাম পাশে যতগুলো ভেরিয়েবল থাকবে, ডান পাশে ঠিক ততগুলোই মান থাকতে হবে; নাহলে `ValueError` দেখাবে।
- এছাড়া একই ভেরিয়েবলে নতুন মান অ্যাসাইন করলে আগের মানটি পরিবর্তিত (Overwrite) হয়ে নতুন মান সংরক্ষিত হয় (যেমন: নিচের কোডে `country`-এর মান `'Finland'` থেকে পরিবর্তিত হয়ে `'Helsink'` হয়েছে)।

```python
# Declaring multiple variables in one line

first_name, last_name, country, age, is_married = 'Asabeneh', 'Yetayeh', 'Helsink', 250, True

print(first_name, last_name, country, age, is_married)
print('First name:', first_name)
print('Last name: ', last_name)
print('Country: ', country)
print('Age: ', age)
print('Married: ', is_married)
```

|---|---|---|---|
| `first_name` | `'Asabeneh'` | `str` | `8` |

| `skills` | `['HTML', 'CSS', 'JS', 'React', 'Python']` | `list` | `5` (মোট উপাদান) |
| `person_info` | `{'firstname': 'Asabeneh', ...}` | `dict` | `4` (মোট key-value জোড়া) |
