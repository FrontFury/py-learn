## ১. স্ট্রিং তৈরি ও সংযোগ (Creating Strings & Concatenation)

Python-এ সিঙ্গেল কোট (`'...'`) অথবা ডাবল কোট (`"..."`)-এর ভেতরে যা লেখা হয় তা **String (`str`)** হিসেবে বিবেচিত হয়।
- **Multiline String:** একাধিক লাইনের প্যারাগ্রাফ বা টেক্সট লিখতে ট্রিপল সিঙ্গেল কোট (`'''...'''`) অথবা ট্রিপল ডাবল কোট (`"""..."""`) ব্যবহৃত হয়।
- **String Concatenation:** `+` অপারেটর ব্যবহার করে একাধিক স্ট্রিংকে জোড়া লাগানো যায়।
- **`len()` ফাংশন:** স্ট্রিংয়ে স্পেসসহ মোট কয়টি ক্যারেক্টার আছে তা গণনা করে।

```python
# Single line comment
letter = 'P'                # A string could be a single character or a bunch of texts
print(letter)               # P
print(len(letter))          # 1
greeting = 'Hello, World!'  # String could be a single or double quote, "Hello, World!"
print(greeting)             # Hello, World!
print(len(greeting))        # 13
sentence = "I hope you are enjoying 30 days of python challenge"
print(sentence)

# Multiline String
multiline_string = '''I am a teacher and enjoy teaching.
I didn't find anything as rewarding as empowering people.
That is why I created 30 days of python.'''
print(multiline_string)

# Another way of doing the same thing
multiline_string = """I am a teacher and enjoy teaching.
I didn't find anything as rewarding as empowering people.
That is why I created 30 days of python."""
print(multiline_string)

# String Concatenation
first_name = 'Asabeneh'
last_name = 'Yetayeh'
space = ' '
full_name = first_name + space + last_name
print(full_name)  # Asabeneh Yetayeh

# Checking length of a string using len() builtin function
print(len(first_name))  # 8
print(len(last_name))   # 7
print(len(first_name) > len(last_name))  # True
print(len(full_name))  # 15
```

## ২. ক্যারেক্টার আনপ্যাকিং, ইনডেক্সিং ও স্লাইসিং (Unpacking, Indexing & Slicing)

- **Unpacking Characters:** স্ট্রিং একটি সিকোয়েন্স হওয়ায় এর প্রতিটি অক্ষরকে আলাদা আলাদা ভেরিয়েবলে অ্যাসাইন করা যায় (যেমন: `a, b, c, d, e, f = 'Python'`)।
- **Positive Indexing:** বাম দিক থেকে শুরু হয় এবং প্রথম অক্ষরের ইনডেক্স **`0`** হয় (যেমন: `'Python'[0]` $\rightarrow$ `'P'`)।
- **Negative Indexing:** ডান দিক (শেষ) থেকে শুরু হয় এবং শেষ অক্ষরের ইনডেক্স **`-1`** হয় (যেমন: `'Python'[-1]` $\rightarrow$ `'n'`)।
- **Slicing (`[start:end:step]`):** স্ট্রিংয়ের নির্দিষ্ট অংশ কেটে নেওয়ার জন্য ব্যবহৃত হয়। এখানে `start` অন্তর্ভুক্ত থাকে, কিন্তু `end` ইনডেক্সের ঠিক আগের অক্ষর পর্যন্ত নেওয়া হয়। তৃতীয় প্যারামিটার `step` দিয়ে নির্দিষ্ট সংখ্যক অক্ষর বাদ দিয়ে (skip করে) নেওয়া যায়।

```python
# Unpacking characters
language = 'Python'
a, b, c, d, e, f = language  # unpacking sequence characters into variables
print(a)  # P
print(b)  # y
print(c)  # t
print(d)  # h
print(e)  # o
print(f)  # n

# Accessing characters in strings by index
language = 'Python'
first_letter = language[0]
print(first_letter)  # P
second_letter = language[1]
print(second_letter)  # y
last_index = len(language) - 1
last_letter = language[last_index]
print(last_letter)  # n

# Negative indexing (-1 is the last index)
language = 'Python'
last_letter = language[-1]
print(last_letter)  # n
second_last = language[-2]
print(second_last)  # o

# Slicing
language = 'Python'
first_three = language[0:3]  # starts at 0 up to 3 (not including 3) -> 'Pyt'
last_three = language[3:6]   # 'hon'
print(last_three)  # hon
# Another way
last_three = language[-3:]
print(last_three)  # hon
last_three = language[3:]
print(last_three)  # hon

# Skipping character while slicing Python strings
language = 'Python'
pto = language[0:6:2]
print(pto)  # Pto
```

## ৩. এসকেপ সিকোয়েন্স (Escape Sequences)

স্ট্রিংয়ের ভেতরে বিশেষ ফরম্যাটিং বা চিহ্ন ব্যবহার করার জন্য ব্যাকস্ল্যাশ (`\`) দিয়ে **Escape Sequence** লেখা হয়:
- `\n` : নতুন লাইন (New line / Line break) তৈরি করে।
- `\t` : একটি ট্যাব স্পেস (Tab space, সাধারণত ৮ স্পেস) তৈরি করে।
- `\\` : স্ট্রিংয়ের ভেতরে একটি ব্যাকস্ল্যাশ (`\`) প্রিন্ট করে।
- `\"` বা `\'` : কোটেশনের ভেতরে ডাবল বা সিঙ্গেল কোট প্রিন্ট করে।

```python
# Escape sequence
print('I hope every one enjoying the python challenge.\nDo you ?')  # line break
print('Days\tTopics\tExercises')
print('Day 1\t3\t5')
print('Day 2\t3\t5')
print('Day 3\t3\t5')
print('Day 4\t3\t5')
print('This is a back slash  symbol (\\)')  # To write a back slash
print('In every programming language it starts with \"Hello, World!\"')
```

## ৪. স্ট্রিং ফরম্যাটিং ও সার্চিং মেথডসমূহ (Formatting & Searching Methods)

- **`capitalize()`:** স্ট্রিংয়ের একদম প্রথম অক্ষরটি বড় হাতের (Capital) করে এবং বাকিগুলো ছোট হাতের রাখে।
- **`title()`:** স্ট্রিংয়ের প্রতিটি শব্দের প্রথম অক্ষর বড় হাতের করে।
- **`swapcase()`:** ছোট হাতের অক্ষরকে বড় হাতের এবং বড় হাতের অক্ষরকে ছোট হাতের করে।
- **`count(substring, start, end)`:** নির্দিষ্ট সাবস্ট্রিংটি স্ট্রিংয়ের ভেতরে কতবার আছে তা গণনা করে।
- **`find(substring)` / `index(substring)`:** সাবস্ট্রিংটি প্রথম কোন ইনডেক্সে পাওয়া গেছে তা রিটার্ন করে। *(পার্থক্য: সাবস্ট্রিং না পাওয়া গেলে `find()` রিটার্ন করে `-1`, কিন্তু `index()` এরর (`ValueError`) দেয়)*।
- **`startswith()` / `endswith()`:** স্ট্রিংটি নির্দিষ্ট শব্দ দিয়ে শুরু বা শেষ হয়েছে কি না তা যাচাই করে (`True`/`False`)।
- **`expandtabs(tabsize)`:** `\t` ক্যারেক্টারকে নির্দিষ্ট সংখ্যক স্পেসে রূপান্তর করে (ডিফল্ট সাইজ `8`)।
- **`format()`:** স্ট্রিংয়ের `{}` প্লেসহোল্ডারের ভেতরে ভেরিয়েবলের মান বসিয়ে সুন্দরভাবে আউটপুট সাজায়।

> **কোড নোট (Bug Fix in Original Snippet):** মূল কোডের বৃত্তের ক্ষেত্রফল বের করার লাইনে `area = pi  # radius ## 2` কমেন্ট আউট হয়ে ছিল, যার ফলে ফলাফল `3.14` আসত। সঠিক সূত্র `area = pi * radius ** 2` ব্যবহার করলে ক্ষেত্রফল `314.0` আসে (নিচের কোডে সঠিক সূত্রটিই দেওয়া হয়েছে)।

```python
# capitalize(): Converts the first character of the string to Capital Letter
challenge = 'thirty days of python'
print(challenge.capitalize())  # 'Thirty days of python'

# count(): returns occurrences of substring in string, count(substring, start=.., end=..)
challenge = 'thirty days of python'
print(challenge.count('y'))        # 3
print(challenge.count('y', 7, 14)) # 1
print(challenge.count('th'))       # 2

# endswith(): Checks if a string ends with a specified ending
challenge = 'thirty days of python'
print(challenge.endswith('on'))    # True
print(challenge.endswith('tion'))  # False

# expandtabs(): Replaces tab character with spaces, default tab size is 8
challenge = 'thirty\tdays\tof\tpython'
print(challenge.expandtabs())      # 'thirty  days    of      python'
print(challenge.expandtabs(10))    # 'thirty    days      of        python'

# find() & index(): Returns the index of first occurrence of substring
challenge = 'thirty days of python'
print(challenge.find('y'))         # 5
print(challenge.find('th'))        # 0
print(challenge.index('y'))        # 5

# format(): formats string into nicer output
first_name = 'Asabeneh'
last_name = 'Yetayeh'
job = 'teacher'
country = 'Finland'
sentence = 'I am {} {}. I am a {}. I live in {}.'.format(first_name, last_name, job, country)
print(sentence)

radius = 10
pi = 3.14
area = pi * radius ** 2
result = 'The area of circle with {} is {}'.format(str(radius), str(area))
print(result)
```

## ৫. স্ট্রিং যাচাইকরণ মেথডসমূহ (Validation / `is...()` Methods)

স্ট্রিংয়ের ভেতরের ক্যারেক্টারগুলোর ধরন পরীক্ষা করার জন্য এই মেথডগুলো সবসময় **`True`** অথবা **`False`** রিটার্ন করে:

- **`isalnum()`:** সব ক্যারেক্টার অক্ষর বা সংখ্যা (Alphanumeric) হলে `True` দেয় (স্পেস থাকলে `False` দেয়)।
- **`isalpha()`:** সব ক্যারেক্টার কেবল বর্ণমালার অক্ষর (Alphabet) হলে `True` দেয় (স্পেস বা সংখ্যা থাকলে `False` দেয়)।
- **`isdecimal()` / `isdigit()` / `isnumeric()`:** স্ট্রিংটি পূর্ণসংখ্যা বা সংখ্যাসূচক ক্যারেক্টার হলে `True` দেয় (দশমিক বিন্দু `.` থাকলে `False` দেয়)।
- **`isidentifier()`:** স্ট্রিংটি Python-এর বৈধ ভেরিয়েবল নাম (Valid Identifier) হওয়ার যোগ্য কি না তা যাচাই করে (সংখ্যা দিয়ে শুরু হলে `False`)।
- **`islower()` / `isupper()`:** স্ট্রিংয়ের সব অক্ষর যথাক্রমে ছোট হাতের (Lowercase) বা বড় হাতের (Uppercase) কি না তা পরীক্ষা করে।

```python
# isalnum(): Checks alphanumeric character
challenge = 'ThirtyDaysPython'
print(challenge.isalnum())  # True

challenge = '30DaysPython'
print(challenge.isalnum())  # True

challenge = 'thirty days of python'
print(challenge.isalnum())  # False (spaces are not alphanumeric)

# isalpha(): Checks if all characters are alphabets
challenge = 'thirty days of python'
print(challenge.isalpha())  # False (because of spaces)
print('ThirtyDaysPython'.isalpha())  # True
num = '123'
print(num.isalpha())        # False

# isdigit(), isdecimal(), isnumeric(): Checks numeric/digit characters
print('Thirty'.isdigit())   # False
print('30'.isdigit())       # True
print('10'.isdecimal())     # True
print('10.5'.isdecimal())   # False
print('10'.isnumeric())     # True
print('ten'.isnumeric())    # False

# isidentifier(): Checks for valid variable name
print('30DaysOfPython'.isidentifier())        # False (starts with a number)
print('thirty_days_of_python'.isidentifier()) # True

# islower() & isupper(): Checks case
print('thirty days of python'.islower())      # True
print('Thirty days of python'.islower())      # False
print('thirty days of python'.isupper())      # False
print('THIRTY DAYS OF PYTHON'.isupper())      # True
```

## ৬. স্ট্রিং পরিবর্তন ও বিভক্তকরণ মেথডসমূহ (`join`, `strip`, `replace`, `split`, `title`, `swapcase`, `startswith`)

- **`join(iterable)`:** একটি লিস্ট বা সিকোয়েন্সের উপাদানগুলোকে নির্দিষ্ট সেপারেটর দিয়ে যুক্ত করে একটি একক স্ট্রিং তৈরি করে।
- **`strip(chars)`:** স্ট্রিংয়ের শুরু (leading) এবং শেষ (trailing) থেকে অপ্রয়োজনীয় স্পেস বা নির্দিষ্ট ক্যারেক্টার মুছে ফেলে। *(নোট: `' thirty days of python '.strip('y')` দিলে শুরু ও শেষে স্পেস থাকায় `'y'` কাটা যাবে না; শুধু `.strip()` দিলে দুই পাশের স্পেস মুছে যাবে)*।
- **`replace(old, new)`:** স্ট্রিংয়ের ভেতরের নির্দিষ্ট শব্দ বা অংশকে নতুন শব্দ দিয়ে প্রতিস্থাপন করে।
- **`split()`:** স্ট্রিংকে স্পেস (বা নির্দিষ্ট সেপারেটর) অনুযায়ী ভাগ করে একটি **List** রিটার্ন করে।

```python
# join(): Returns a concatenated string
web_tech = ['HTML', 'CSS', 'JavaScript', 'React']
result = '# '.join(web_tech)
print(result)  # 'HTML# CSS# JavaScript# React'

# strip(): Removes both leading and trailing characters/spaces
challenge = ' thirty days of python '
print(challenge.strip())     # 'thirty days of python'
print('yythirty days of pythonyy'.strip('y'))  # 'thirty days of python'

# replace(): Replaces substring inside
challenge = 'thirty days of python'
print(challenge.replace('python', 'coding'))  # 'thirty days of coding'

# split(): Splits String into a list
challenge = 'thirty days of python'
print(challenge.split())  # ['thirty', 'days', 'of', 'python']

# title(): Returns a Title Cased String
challenge = 'thirty days of python'
print(challenge.title())  # Thirty Days Of Python

# swapcase(): Swaps uppercase to lowercase and vice versa
print('thirty days of python'.swapcase())  # THIRTY DAYS OF PYTHON
print('Thirty Days Of Python'.swapcase())  # tHIRTY dAYS oF pYTHON

# startswith(): Checks if String Starts with the Specified String
print('thirty days of python'.startswith('thirty'))  # True
print('30 days of python'.startswith('thirty'))      # False
```
