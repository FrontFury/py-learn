## ১. লিস্ট কী এবং কীভাবে তৈরি করতে হয়? (Creating Lists)

**List** হলো Python-এর অন্যতম গুরুত্বপূর্ণ বিল্ট-ইন ডেটা স্ট্রাকচার, যা একটি একক ভেরিয়েবলের ভেতরে একাধিক আইটেম সংরক্ষণ করে।
- লিস্ট **Ordered** (ক্রমবদ্ধ), **Mutable** (পরিবর্তনযোগ্য) এবং এতে ডুপ্লিকেট (একই) মান রাখা যায়।
- **লিস্ট তৈরির ২টি উপায়:**
  1. বিল্ট-ইন `list()` ফাংশন ব্যবহার করে (যেমন: `empty_list = list()`)।
  2. স্কয়ার ব্র্যাকেট `[]` ব্যবহার করে (যেমন: `fruits = ['banana', 'orange', 'mango', 'lemon']`)।
- **`len()` ফাংশন:** লিস্টের ভেতরে মোট কয়টি উপাদান আছে তা গণনা করে।

```python
empty_list = list()  # this is an empty list, no item in the list
print(len(empty_list))  # 0

# list of fruits
fruits = ['banana', 'orange', 'mango', 'lemon']
vegetables = ['Tomato', 'Potato', 'Cabbage', 'Onion', 'Carrot']      # list of vegetables
animal_products = ['milk', 'meat', 'butter', 'yoghurt']             # list of animal products
web_techs = ['HTML', 'CSS', 'JS', 'React', 'Redux', 'Node', 'MongDB']  # list of web technologies
countries = ['Finland', 'Estonia', 'Denmark', 'Sweden', 'Norway']

# Print the lists and their length
print('Fruits:', fruits)
print('Number of fruits:', len(fruits))
print('Vegetables:', vegetables)
print('Number of vegetables:', len(vegetables))
print('Animal products:', animal_products)
print('Number of animal products:', len(animal_products))
print('Web technologies:', web_techs)
print('Number of web technologies:', len(web_techs))
print('Number of countries:', len(countries))
```

## ২. ইনডেক্সিং, স্লাইসিং ও মান পরিবর্তন (Indexing, Slicing & Modifying)

- **Positive Indexing (`0` থেকে শুরু):** প্রথম আইটেম `fruits[0]`, দ্বিতীয় আইটেম `fruits[1]` এবং শেষ আইটেম `fruits[len(fruits) - 1]`।
- **Negative Indexing (`-1` থেকে শুরু):** ডান দিক থেকে শেষ আইটেম `fruits[-1]`, শেষের আগেরটি `fruits[-2]`।
- **Slicing (`[start:end]`):** নির্দিষ্ট রেঞ্জের আইটেমগুলো নিয়ে নতুন একটি সাব-লিস্ট তৈরি করে (`end` ইনডেক্সের আগের আইটেম পর্যন্ত নেয়)।
- **Modifying Items:** লিস্ট পরিবর্তনযোগ্য (Mutable) হওয়ায় ইনডেক্স ধরে সরাসরি মান বদলানো যায় (যেমন: `fruits[0] = 'Avocado'`)।

> **কোড নোট (Bug Fix in Original Snippet):** মূল কোডে শেষ আইটেম পরিবর্তন করতে `last_index = len(fruits)` লেখা হয়েছিল। যেহেতু ৪টি আইটেমের লিস্টে ইনডেক্স `0` থেকে `3` পর্যন্ত থাকে, তাই `fruits[4]` দিলে `IndexError` আসে। সঠিক শেষ ইনডেক্স হলো **`len(fruits) - 1`**।

```python
# Accessing items using Positive Indexing
fruits = ['banana', 'orange', 'mango', 'lemon']
first_fruit = fruits[0]
print(first_fruit)      # banana
second_fruit = fruits[1]
print(second_fruit)     # orange
last_index = len(fruits) - 1
last_fruit = fruits[last_index]
print(last_fruit)       # lemon

# Accessing items using Negative Indexing
last_fruit = fruits[-1]
second_last = fruits[-2]
print(last_fruit)       # lemon
print(second_last)      # mango

# Slicing items
all_fruits = fruits[0:4]          # ['banana', 'orange', 'mango', 'lemon']
all_fruits = fruits[0:]           # ['banana', 'orange', 'mango', 'lemon']
orange_and_mango = fruits[1:3]    # ['orange', 'mango']
orange_mango_lemon = fruits[1:]   # ['orange', 'mango', 'lemon']
print(orange_and_mango)
print(orange_mango_lemon)

# Negative Slicing
all_fruits_neg = fruits[-4:]      # ['banana', 'orange', 'mango', 'lemon']
orange_and_mango_neg = fruits[-3:-1] # ['orange', 'mango']
orange_mango_lemon_neg = fruits[-3:] # ['orange', 'mango', 'lemon']
print(orange_and_mango_neg)

# Modifying list items
fruits = ['banana', 'orange', 'mango', 'lemon']
fruits[0] = 'Avocado'
print(fruits)  # ['Avocado', 'orange', 'mango', 'lemon']
fruits[1] = 'apple'
print(fruits)  # ['Avocado', 'apple', 'mango', 'lemon']
last_index = len(fruits) - 1
fruits[last_index] = 'lime'
print(fruits)  # ['Avocado', 'apple', 'mango', 'lime']
```

## ৩. আইটেম যাচাই ও নতুন আইটেম যোগ করা (`in`, `append`, `insert`)

- **`in` অপারেটর:** লিস্টের ভেতরে নির্দিষ্ট কোনো আইটেম আছে কি না তা যাচাই করে (`True` অথবা `False` রিটার্ন করে)।
- **`append(item)`:** লিস্টের একদম **শেষে** একটি নতুন আইটেম যোগ করে।
- **`insert(index, item)`:** লিস্টের যেকোনো নির্দিষ্ট **ইনডেক্সে** নতুন আইটেম প্রবেশ করায় (এবং পরের আইটেমগুলো ডানদিকে সরে যায়)। *(নোট: মূল কোডে ভুলে `fruits.list(3, 'lime')` লেখা ছিল; সঠিক মেথডটি হলো `fruits.insert(3, 'lime')`)*।

```python
# Checking items in a list
fruits = ['banana', 'orange', 'mango', 'lemon']
does_exist = 'banana' in fruits
print(does_exist)  # True
does_exist = 'lime' in fruits
print(does_exist)  # False

# Append: adds item to the end of the list
fruits = ['banana', 'orange', 'mango', 'lemon']
fruits.append('apple')
print(fruits)      # ['banana', 'orange', 'mango', 'lemon', 'apple']
fruits.append('lime')
print(fruits)      # ['banana', 'orange', 'mango', 'lemon', 'apple', 'lime']

# Insert: inserts item at a specified index
fruits = ['banana', 'orange', 'mango', 'lemon']
fruits.insert(2, 'apple')  # insert apple between orange and mango
print(fruits)              # ['banana', 'orange', 'apple', 'mango', 'lemon']
fruits.insert(3, 'lime')   # insert lime at index 3
print(fruits)              # ['banana', 'orange', 'apple', 'lime', 'mango', 'lemon']
```

## ৪. লিস্ট থেকে আইটেম মুছে ফেলা (`remove`, `pop`, `del`, `clear`)

Python লিস্ট থেকে আইটেম বাদ দেওয়ার চারটি প্রধান পদ্ধতি রয়েছে:
1. **`remove(item)`:** আইটেমের **নাম বা মান** ধরে প্রথম ম্যাচটি মুছে ফেলে।
2. **`pop(index)`:** নির্দিষ্ট ইনডেক্সের আইটেম মুছে ফেলে এবং সেটি রিটার্ন করে। কোনো ইনডেক্স না দিলে (`fruits.pop()`) একদম **শেষের আইটেমটি** মুছে ফেলে। *(নোট: মূল স্নিপেটে ভুলে `fruits.remove()` ও `fruits.remove(0)` লেখা ছিল; ইনডেক্স দিয়ে মুছতে `pop()` ব্যবহৃত হয়)*।
3. **`del` স্টেটমেন্ট:** নির্দিষ্ট ইনডেক্সের আইটেম (`del fruits[0]`) অথবা সম্পূর্ণ লিস্ট ভেরিয়েবলটিই মেমোরি থেকে ডিলিট (`del fruits`) করে দেয়। সম্পূর্ণ লিস্ট ডিলিট করার পর প্রিন্ট করতে চাইলে `NameError` দেখাবে।
4. **`clear()`:** লিস্টের ভেতরের সব আইটেম মুছে লিস্টটিকে ফাঁকা (`[]`) করে দেয়, তবে লিস্ট ভেরিয়েবলটি মেমোরিতে থেকে যায়।

```python
# remove(): removes a specified item by value
fruits = ['banana', 'orange', 'mango', 'lemon']
fruits.remove('banana')
print(fruits)  # ['orange', 'mango', 'lemon']
fruits.remove('lemon')
print(fruits)  # ['orange', 'mango']

# pop(): removes item at specified index (or last item if index not specified)
fruits = ['banana', 'orange', 'mango', 'lemon']
fruits.pop()       # removes last item ('lemon')
print(fruits)      # ['banana', 'orange', 'mango']
fruits.pop(0)      # removes item at index 0 ('banana')
print(fruits)      # ['orange', 'mango']

# del: removes item by index or deletes the entire list
fruits = ['banana', 'orange', 'mango', 'lemon']
del fruits[0]
print(fruits)      # ['orange', 'mango', 'lemon']
del fruits[1]
print(fruits)      # ['orange', 'lemon']
del fruits         # deletes the variable completely (printing it would raise NameError)

# clear(): empties the list
fruits = ['banana', 'orange', 'mango', 'lemon']
fruits.clear()
print(fruits)      # []
```

## ৫. কপি, জয়েন, গণনা, রিভার্স ও সর্টিং (`copy`, `+`, `extend`, `count`, `index`, `reverse`, `sort`)

- **`copy()`:** মূল লিস্টের একটি স্বাধীন কপি তৈরি করে (যাতে কপি পরিবর্তন করলে মূল লিস্ট পরিবর্তিত না হয়)।
- **`+` অপারেটর বনাম `extend()`:** `+` অপারেটর দুটি লিস্ট জোড়া লাগিয়ে একটি **নতুন লিস্ট** তৈরি করে; অন্যদিকে `list1.extend(list2)` সরাসরি `list1`-এর ভেতরেই `list2`-এর উপাদানগুলো যুক্ত করে দেয়।
- **`count(item)`:** লিস্টে নির্দিষ্ট আইটেমটি মোট কতবার আছে তা গণনা করে।
- **`index(item)`:** আইটেমটি লিস্টে প্রথম কোন ইনডেক্সে আছে তা রিটার্ন করে।
- **`reverse()`:** লিস্টের ক্রম উল্টে দেয়। *(নোট: `fruits.reverse()` সরাসরি মূল লিস্টটি পরিবর্তন (In-place) করে এবং `None` রিটার্ন করে; তাই `print(fruits.reverse())` করলে `None` প্রিন্ট হয়। উল্টানো লিস্ট দেখতে হলে আগে `fruits.reverse()` কল করে তারপর `print(fruits)` লিখতে হয়)*।
- **`sort()` / `sort(reverse=True)`:** লিস্টের উপাদানগুলোকে ছোট থেকে বড় (Ascending) বা বড় থেকে ছোট (Descending) ক্রমে সাজায়।

```python
# Copying a list
fruits = ['banana', 'orange', 'mango', 'lemon']
fruits_copy = fruits.copy()
print(fruits_copy)  # ['banana', 'orange', 'mango', 'lemon']

# Joining lists using '+' operator
positive_numbers = [1, 2, 3, 4, 5]
zero = [0]
negative_numbers = [-5, -4, -3, -2, -1]
integers = negative_numbers + zero + positive_numbers
print(integers)

fruits = ['banana', 'orange', 'mango', 'lemon']
vegetables = ['Tomato', 'Potato', 'Cabbage', 'Onion', 'Carrot']
fruits_and_vegetables = fruits + vegetables
print(fruits_and_vegetables)

# Joining lists using extend()
num1 = [0, 1, 2, 3]
num2 = [4, 5, 6]
num1.extend(num2)
print('Numbers:', num1)

negative_numbers = [-5, -4, -3, -2, -1]
positive_numbers = [1, 2, 3, 4, 5]
zero = [0]
negative_numbers.extend(zero)
negative_numbers.extend(positive_numbers)
print('Integers:', negative_numbers)

# count() & index()
fruits = ['banana', 'orange', 'mango', 'lemon']
print(fruits.count('orange'))   # 1
print(fruits.index('orange'))   # 1

ages = [22, 19, 24, 25, 26, 24, 25, 24]
print(ages.count(24))           # 3
print(ages.index(24))           # 2 (first occurrence)

# reverse() - modifies list in-place and returns None
fruits = ['banana', 'orange', 'mango', 'lemon']
fruits.reverse()
print('Reversed fruits:', fruits)

ages = [22, 19, 24, 25, 26, 24, 25, 24]
ages.reverse()
print('Reversed ages:', ages)

# sort() - ascending and descending
fruits = ['banana', 'orange', 'mango', 'lemon']
fruits.sort()
print('Sorted fruits (asc):', fruits)
fruits.sort(reverse=True)
print('Sorted fruits (desc):', fruits)

ages = [22, 19, 24, 25, 26, 24, 25, 24]
ages.sort()
print('Sorted ages (asc):', ages)
ages.sort(reverse=True)
print('Sorted ages (desc):', ages)
```

## ৬. দ্রুত রিভিশন (Quick Revision Table)

| মেথড / অপারেটর | কাজ | রিটার্ন ভ্যালু / আচরণ |
|---|---|---|
| `append(x)` | লিস্টের শেষে `x` যোগ করে | মূল লিস্ট পরিবর্তন করে (`None`) |
| `insert(i, x)` | নির্দিষ্ট `i` ইনডেক্সে `x` যোগ করে | মূল লিস্ট পরিবর্তন করে (`None`) |
| `remove(x)` | প্রথম ম্যাচ হওয়া `x` মানটি মুছে ফেলে | মূল লিস্ট পরিবর্তন করে (`None`) |
| `pop(i)` | `i` ইনডেক্সের (বা শেষের) আইটেম মুছে ফেলে | মুছে ফেলা আইটেমটি রিটার্ন করে |
| `clear()` | লিস্টের সব আইটেম মুছে খালি করে দেয় | মূল লিস্ট খালি (`[]`) করে |
| `copy()` | লিস্টের একটি কপি তৈরি করে | নতুন লিস্ট রিটার্ন করে |
| `extend(list2)` | আরেকটি লিস্টের সব আইটেম শেষে যুক্ত করে | মূল লিস্ট পরিবর্তন করে (`None`) |
| `count(x)` | লিস্টে `x` কতবার আছে তা গণনা করে | পূর্ণসংখ্যা (`int`) রিটার্ন করে |
| `index(x)` | `x`-এর প্রথম অবস্থান (ইনডেক্স) বের করে | পূর্ণসংখ্যা (`int`) রিটার্ন করে |
| `reverse()` | লিস্টের ক্রম উল্টে দেয় | মূল লিস্ট পরিবর্তন করে (`None`) |
| `sort()` | লিস্টকে ছোট থেকে বড় ক্রমে সাজায় | মূল লিস্ট পরিবর্তন করে (`None`) |
