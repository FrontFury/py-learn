## ১. গাণিতিক অপারেটর (Arithmetic Operators) ও সংখ্যার ধরন

Python-এ পূর্ণসংখ্যা (`int`), দশমিক সংখ্যা (`float`) এবং জটিল সংখ্যা (`complex`)-এর ওপর বিভিন্ন গাণিতিক অপারেশন করা যায়:

- **Division (`/`) বনাম Floor Division (`//`):** সাধারণ ভাগ (`/`) সবসময় দশমিক (`float`) সংখ্যা রিটার্ন করে (যেমন: `4 / 2` $\rightarrow$ `2.0`), কিন্তু Floor Division (`//`) দশমিকের পরের অংশ বাদ দিয়ে পূর্ণসংখ্যা দেয় (যেমন: `7 // 2` $\rightarrow$ `3`)।
- **Modulus (`%`):** ভাগশেষ (Remainder) নির্ণয় করে (যেমন: `3 % 2` $\rightarrow$ `1`)।
- **Exponential (`**`):** ঘাত বা পাওয়ার হিসাব করে (যেমন: `3 ** 2` $= 3^2 = 9$)।
- **Complex Numbers:** জটিল সংখ্যার কাল্পনিক অংশ বোঝাতে Python-এ `j` ব্যবহার করা হয়। যেমন: `(1 + 1j) * (1 - 1j)` গুণ করলে ফলাফল আসে `(2+0j)`।

```python
# Arithmetic Operations in Python - Day 3
# Integers

print('Addition: ', 1 + 2)
print('Subtraction: ', 2 - 1)
print('Multiplication: ', 2 * 3)
# Division in python gives floating number
print('Division: ', 4 / 2)
print('Division: ', 6 / 2)
print('Division: ', 7 / 2)
# gives without the floating number or without the remaining
print('Division without the remainder: ', 7 // 2)
print('Modulus: ', 3 % 2)                           # Gives the remainder
print('Division without the remainder: ', 7 // 3)
print('Exponential: ', 3 ** 2)                     # it means 3 * 3

# Floating numbers
print('Floating Number,PI', 3.14)
print('Floating Number, gravity', 9.81)

# Complex numbers
print('Complex number: ', 1+1j)
print('Multiplying complex number: ', (1+1j) * (1-1j))
```

## ২. ভেরিয়েবল ব্যবহার করে গাণিতিক হিসাব (Variables & Arithmetic)

সরাসরি সংখ্যার বদলে ভেরিয়েবলে মান রেখে গাণিতিক অপারেশন করলে কোড আরও পরিচ্ছন্ন ও পুনর্ব্যবহারযোগ্য হয়:

- **Built-in Function এড়িয়ে চলা:** যোগফল রাখার জন্য `sum`-এর পরিবর্তে `total` ব্যবহার করা উত্তম, কারণ `sum()` Python-এর একটি বিল্ট-ইন ফাংশন। বিল্ট-ইন ফাংশনের নামে ভেরিয়েবল ডিক্লেয়ার করলে মূল ফাংশনটি ওভাররাইড (Override) হয়ে যায়।
- **Label সহ প্রিন্ট:** `print()`-এর ভেতরে বর্ণনামূলক টেক্সট (Label) দিলে আউটপুটের কোন মানটি কিসের তা সহজেই বোঝা যায়।

```python
# Declaring the variable at the top first

a = 3  # a is a variable name and 3 is an integer data type
b = 2  # b is a variable name and 2 is an integer data type

# Arithmetic operations and assigning the result to a variable
total = a + b
diff = a - b
product = a * b
division = a / b
remainder = a % b
floor_division = a // b
exponential = a ** b

# I should have used sum instead of total but sum is a built-in function try to avoid overriding builtin functions
print(total)  # if you don't label your print with some string, you never know from where the result is coming
print('a + b = ', total)
print('a - b = ', diff)
print('a * b = ', product)
print('a / b = ', division)
print('a % b = ', remainder)
print('a // b = ', floor_division)
print('a ** b = ', exponential)

# Declaring values and organizing them together
num_one = 3
num_two = 4

# Arithmetic operations
total = num_one + num_two
diff = num_two - num_one
product = num_one * num_two
div = num_two / num_two
remainder = num_two % num_one

# Printing values with label
print('total: ', total)
print('difference: ', diff)
print('product: ', product)
print('division: ', div)
print('remainder: ', remainder)
```

## ৩. বাস্তবভিত্তিক গাণিতিক সূত্রের প্রয়োগ (Practical Formulas)

Python-এর অপারেটর ব্যবহার করে জ্যামিতি ও পদার্থবিজ্ঞানের সূত্র সহজেই সমাধান করা যায়:
- **বৃত্তের ক্ষেত্রফল (Area of a Circle):** $\text{Area} = \pi r^2$ (`3.14 * radius ** 2`)
- **আয়তক্ষেত্রের ক্ষেত্রফল (Area of a Rectangle):** $\text{Area} = \text{length} \times \text{width}$ (`length * width`)
- **বস্তুর ওজন (Weight):** $W = m \times g$ (`mass * gravity`)

```python
# Calculating area of a circle
radius = 10                                 # radius of a circle
# two * sign means exponent or power
area_of_circle = 3.14 * radius ** 2
print('Area of a circle:', area_of_circle)

# Calculating area of a rectangle
length = 10
width = 20
area_of_rectangle = length * width
print('Area of rectangle:', area_of_rectangle)

# Calculating a weight of an object
mass = 75
gravity = 9.81
weight = mass * gravity
print(weight, 'N')
```

## ৪. তুলনামূলক অপারেটর (Comparison Operators)

দুটি মান বা এক্সপ্রেশনের মধ্যে তুলনা করার জন্য **Comparison Operators** ব্যবহৃত হয় এবং এগুলো সবসময় বুলিয়ান মান (**`True`** অথবা **`False`**) রিটার্ন করে:

| অপারেটর | নাম | উদাহরণ | ফলাফল |
|---|---|---|---|
| `==` | Equal (সমান) | `3 == 2` | `False` |
| `!=` | Not Equal (সমান নয়) | `3 != 2` | `True` |
| `>` | Greater than (বড়) | `3 > 2` | `True` |
| `<` | Less than (ছোট) | `3 < 2` | `False` |
| `>=` | Greater than or equal to (বড় বা সমান) | `3 >= 2` | `True` |
| `<=` | Less than or equal to (ছোট বা সমান) | `2 <= 3` | `True` |

> **স্ট্রিং-এর দৈর্ঘ্য তুলনা (`len()`):** `len('mango')` হলো `5` এবং `len('avocado')` হলো `7`। তাই `len('mango') < len('avocado')` সত্য (`True`)।

```python
print(3 > 2)     # True, because 3 is greater than 2
print(3 >= 2)    # True, because 3 is greater than or equal to 2
print(3 < 2)     # False, because 3 is greater than 2
print(2 < 3)     # True, because 2 is less than 3
print(2 <= 3)    # True, because 2 is less than or equal to 3
print(3 == 2)    # False, because 3 is not equal to 2
print(3 != 2)    # True, because 3 is not equal to 2
print(len('mango') == len('avocado'))  # False (5 == 7)
print(len('mango') != len('avocado'))  # True  (5 != 7)
print(len('mango') < len('avocado'))   # True  (5 < 7)
print(len('milk') != len('meat'))      # False (4 != 4)
print(len('milk') == len('meat'))      # True  (4 == 4)
print(len('tomato') == len('potato'))  # True  (6 == 6)
print(len('python') > len('dragon'))   # False (6 > 6)
```

## ৫. আইডেন্টিটি (`is`, `is not`), মেম্বারশিপ (`in`) এবং লজিক্যাল অপারেটর (`and`, `or`, `not`)

### ১. Identity Operators (`is`, `is not`)
- দুটি ভেরিয়েবল মেমোরিতে একই অবজেক্টকে নির্দেশ করছে কি না তা পরীক্ষা করে। যেমন: `1 is 1` $\rightarrow$ `True` এবং `1 is not 2` $\rightarrow$ `True`। *(নোট: মান সমান কি না তা যাচাই করতে সাধারণত `==` ব্যবহার করা উত্তম)*।

### ২. Membership Operators (`in`, `not in`)
- কোনো স্ট্রিং, লিস্ট বা সিকোয়েন্সের ভেতরে নির্দিষ্ট কোনো উপাদান বা সাবস্ট্রিং আছে কি না তা যাচাই করে।
- `'A' in 'Asabeneh'` $\rightarrow$ `True`, কিন্তু `'B' in 'Asabeneh'` $\rightarrow$ `False` (কারণ `'Asabeneh'`-এ বড় হাতের `'B'` নেই; এটি Case-sensitive)।

### ৩. Logical Operators (`and`, `or`, `not`)
- **`and`:** উভয় শর্ত `True` হলেই কেবল `True` দেয়, নাহলে `False`।
- **`or`:** যেকোনো একটি শর্ত `True` হলেই `True` দেয়; উভয় শর্ত `False` হলে `False` দেয়।
- **`not`:** বুলিয়ান ফলাফলকে উল্টে দেয় (`not True` $\rightarrow$ `False` এবং `not False` $\rightarrow$ `True`)।

```python
# Boolean comparison
print('True == True: ', True == True)
print('True == False: ', True == False)
print('False == False:', False == False)
print('True and True: ', True and True)
print('True or False:', True or False)

# Another way comparison
print('1 is 1:', 1 == 1)                  # True - values are the same
print('1 is not 2:', 1 != 2)              # True - 1 is not 2
print('A in Asabeneh:', 'A' in 'Asabeneh')  # True - A found in the string
print('B in Asabeneh:', 'B' in 'Asabeneh')  # False - there is no uppercase B
print('coding in coding for all:', 'coding' in 'coding for all')  # True
print('a in an:', 'a' in 'an')            # True
print('4 is 2 ** 2:', 4 == 2 ** 2)        # True

# Logical Operators
print(3 > 2 and 4 > 3)  # True - because both statements are true
print(3 > 2 and 4 < 3)  # False - because the second statement is false
print(3 < 2 and 4 < 3)  # False - because both statements are false
print(3 > 2 or 4 > 3)   # True - because both statements are true
print(3 > 2 or 4 < 3)   # True - because one of the statements is true
print(3 < 2 or 4 < 3)   # False - because both statements are false
print(not 3 > 2)        # False - because 3 > 2 is true, then not True gives False
print(not True)         # False - Negation, the not operator turns true to false
print(not False)        # True
print(not not True)     # True
print(not not False)    # False
```
